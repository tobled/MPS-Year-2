So for loops are really powerful but what happens if we don't know how many times we want to execute a command

If for example we want to execute a command until a condition is met, well this is the while loop.

```python
x = 10
while x > 0:
    print(x)
    x = x - 1
```

The code indented above was executed until the condition was no longer true, now what your probably thinking is that the above code could easily be done in a for loop, well yes but the code below would be difficult

```python
import math # this gives me access to some maths functions
x = 0
delta = 0.01 # try making this a smaller number
condition = math.sin(x) < math.cos(x)
while condition:
    x = x + delta
    condition = math.sin(x) < math.cos(x)
print('The radians at which cos and sin are similar are ', x)
print("The value of sin at that point is", math.sin(x),"and cos is", math.cos(x))
```

## A Python Integer Is More Than Just an Integer

The standard Python implementation is written in C.
This means that every Python object is simply a cleverly-disguised C structure, which contains not only its value, but other information as well. For example, when we define an integer in Python, such as ``x = 10000``, ``x`` is not just a "raw" integer. It's actually a pointer to a compound C structure, which contains several values.


A single integer in Python 3.12 actually contains four pieces:

- ``ob_refcnt``, a reference count that helps Python silently handle memory allocation and deallocation
- ``ob_type``, which encodes the type of the variable
- ``ob_size``, which specifies the size of the following data members
- ``ob_digit``, which contains the actual integer value that we expect the Python variable to represent.

This means that there is some overhead in storing an integer in Python as compared to an integer in a compiled language like C.

Hidden with in each of definition types are methods. These are functions which are associated with that datatype. Lets have a look at an example.

```python
myInteger = 2
print(myInteger.__sizeof__()) # The .__sizeof__() method tell me how much space my variable takes up.
print(myInteger.bit_length()) # This tells me how many bits I need to store the data
print('The sizeof method returned 28 bytes, which is 224 bits')
```

Let's try some methods with floats.

```python
myFloat = 30.1
print(myFloat.__sizeof__())
print(myFloat.hex()) # In hexadecimal format
try:
    print(myInteger.hex()) # This doesn't work because an integer does not have the .hex() code in its definition.
except:
    print('myInteger.hex() failed')
```

The error message above is really interesting, what it tells you is that there is very different code behind the float and interger data types.

### Methods
These built in functions we have been using are called methods. We don't really care if we use a method or a function, they both do the same thing.
Here is an example of using a method and a function on the same data type. Notice how the function returns a value but the method operates on the data.

```python
# Function
import numpy as np
data = np.asarray([6,5,4,3,2,1])
data = np.sort(data)
print(data)
```

```python
# Method
import numpy as np
data = np.asarray([6,5,4,3,2,1])
data.sort()
print(data)
```

For loops, the foor loop is a very valuable tool for performing an operation a set number of time.

The code below looks similar to the IF statement, the indented code is the code to be executed over and over again.  The for line, is very special.

for tells python the it id going to perform something a number of times, in this case 10 times, actually the for command is expecting a list of items to execute (I will explain this more later) for now we use the range() command to feed the for operator the numbers 0--> 9

```python
for I in range(10):
    print(I)
```

Ok, so I said that the for command expects a list of items to execute, the code below demonstrates this.

```python
myAnimals = ['Fox', 'Cat', 'Dog', 'Bunny', 'Aligator', 'Tree']
for animal in myAnimals:
    print(animal)
```

Cool eh, we can put a loop in a loop

```python
myAnimals = ['Fox', 'Cat', 'Dog', 'Bunny', 'Aligator', 'Tree']
for animal in myAnimals:
    for I in range(3):
        print(animal)
```

This is useful if you want to perform an operation a number of time which varies on the number of times the original loop has executed, can you explain why this command performs the output

```python
for I in range(10):
    for X in range(I):
        print('*', end='') # This end command tells the print statement not to start a new line
    print()
```

We are going to look at some python code to work out what it is doing in order to build up some basic functionality

```python
myVariable = 1  # set the variable named myVariable to the value of 1, this will by defalut be set to an int, because the number has not decimal point
print(myVariable)  #shows the variable
print(type(myVariable)) # shows the type of variable
```

Now we can store things lets try doing some maths.

```python
myVariable = 1    # Think of this as storing the number 1 in a box in the computer
myVariable = myVariable + 1  # Now we can add 1 to that box
print(myVariable) # show the value of the variable
```

Over the next few cells you will see some basic maths. Once we have defined our two variables we add them together and store them in a new variable called temp.  We then print the name of the operation, in the first case this is Addition. We then print the value stored in the new variable temp.

We can also just print the answer with out making the new variable, as in the last but one line.

```python
number_1 = 10
number_2 = 5

temp = number_1 + number_2      # Add the two variables together
print("Addition")               # Print statement
print(temp)                     # Print the result of the addition.
print(type(temp))
print(number_1 + number_2)      # Print the addition directly to the console
print(" ")                      # Print a blank line
```

```python
temp = number_1 - number_2
print("Subtraction")
print(temp)
print(type(temp))
print(number_1 - number_2)
print(" ")
```

```python
temp = number_1 * number_2
print("Multiplication")
print(temp)
print(type(temp))
print(number_1 * number_2)
print(" ")
```

# Type of operation

```python
# Addition: Adds two numbers together.
print("10 + 5 =", 10 + 5) # Output: 10 + 5 = 15

# Subtraction: Subtracts the second number from the first.
print("10 - 5 =", 10 - 5) # Output: 10 - 5 = 5

# Multiplication: Multiplies two numbers.
print("10 * 5 =", 10 * 5) # Output: 10 * 5 = 50

# Division: Divides the first number by the second.
# In Python 3, this always returns a float (decimal number).
print("10 / 5 =", 10 / 5) # Output: 10 / 5 = 2.0

# Modulus: Returns the remainder after division.
# For example, 10 divided by 3 is 3 with a remainder of 1.
print("10 % 3 =", 10 % 3) # Output: 10 % 3 = 1

# Exponentiation: Raises the first number to the power of the second number.
# 2 ** 3 means 2 * 2 * 2.
print("2 ** 3 =", 2 ** 3) # Output: 2 ** 3 = 8
```

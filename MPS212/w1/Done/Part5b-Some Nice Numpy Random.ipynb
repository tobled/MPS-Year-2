##A real world example.
A company drills nine wild-cat oil exploration wells, each with an estimated probability of success of 0.1.
Unfortunately, all nine wells fail. What is the probability of that happening?
We use the binomial probability function here.


```python
probabilities = np.random.binomial(50, 0.1, 1)   # The first number is the number of events
                                                # The second is the probability of an outcome (in this case striking oil).
                                                # The final number is the number of times to perform the simulation.
print(probabilities)
# This returns a single outcome
```

```python
numberOfRuns = 20
probabilities = np.random.binomial(9, 0.1, numberOfRuns) # We do this for 20 simulations.
print(probabilities)
```

```python
# Lets only select the times that none of the wells strike oil, i.e. when the output is 0
noStrike = probabilities == 0 # this produces a list of true or false/
print((noStrike))
print('The oil wells all dont strike oil', sum(noStrike), numberOfRuns, 'times')
```

```python
# Now lets do this a lot of time
numberOfRuns = 20000000
probabilities = np.random.binomial(9, 0.1, numberOfRuns) # We do this for 20 simulations.
noStrike = probabilities == 0
print('All the oil wells dont strike oil', sum(noStrike), numberOfRuns, 'times')
print('The probabity of all 9 wells not striking oil is', 100 * sum(noStrike) /numberOfRuns,'%')
```

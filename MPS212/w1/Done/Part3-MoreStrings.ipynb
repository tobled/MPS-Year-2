Playing with strings gives us a great opportunity to explore some methods in python

A method in python is somewhat similar to a function, except it is associated with object/classes. Methods in python are very similar to functions except for two major differences. The method is implicitly used for an object for which it is called. The method is accessible to data that is contained within the class

```python
myString = "Are bunny kittens really that amazing"
```

A method allows you to use the variable as a function so we would write
myString.COMMAND I WANT TO USE(). for example. <Br>
Although we often treat functions and methods as similar operations, we need to understand the actual undelting difference and strings (which have a lot of methods) are great for that.
A function takes a paramater and in general, returns an answer.
A method, which is built in to each instance of the data type acts on that instance, so if I call myvariable.themethod() the answer is stored in myvariable, i.e. it acts on itself. If this makes no sense, ask someone in the lab.

```python
print(myString.capitalize())
print('')
print(myString.lower())
print('')
print(myString.swapcase())
print('')
print(myString.title())
print('')
print(myString.upper())
```

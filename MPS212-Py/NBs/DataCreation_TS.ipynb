```python
import numpy as np
set = ['A', 'C', 'G', 'T']
data = []
for I in range(1000):
    data.append(set[np.random.randint(0,4)])
```

```python
print(data)
```

```python
test = ['A', 'C', 'G', 'T']
for I in range(0,len(data)-4):
    current = data[I:I+4]
    if current == test:
        print(I)
```

```python
import pickle

with open('my_list.pkl', 'wb') as f:
    pickle.dump(data, f)
```

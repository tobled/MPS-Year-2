So we have numbers and strings, lets look at lists
List are really nice ways to store lots of strings or numbers.

First lets define a list of animals

```python
myAnimals = ['Fox', 'Cat', 'Dog', 'Bunny', 'Aligator', 'Tree']
# to build a list of items, you simply supply python a list of comma seperated values between square brackets.
```

```python
print(myAnimals)        # This will print the list
print(myAnimals[0:3])   # Just like strings and numpy arrays we can slice lists here we will print the 0th, 1st and 2nd item.
```

List can hold anything. We have just used a list of names, i.e. strings. However, we can store anything and they (unlike numpy arrays) don't have to be the same data type/

```python
myList = [1, 2, 'Three', 4, 'Five']
print(myList)
```

```python
print(len(myList))
try:
    print(max(myList))
except:
    print('something went wrong')
try:
    print(min(myList))
except:
    print('something went wrong')
```

This breaks because python does not know how to compare strings and ints, however: The code below works, here python treats max and min as alphabetic maximum and minimum.

```python
print(len(myAnimals)) # how long is the list, i,e. how many items.
print(max(myAnimals)) # Which comes last alphabetically
print(min(myAnimals)) # Which comes first alphabetically
```

List can be nested, i.e. a list can contain a list

```python
myNewList = [myList, myAnimals]
print(myNewList)
```

Lists can me manipulates, they are mutable. Here we delete an item

```python
print(myAnimals)
del myAnimals[5]
print(myAnimals)
```

But we can also add an item

```python
print(myAnimals)
myAnimals.append("Frog")
print(myAnimals)
myAnimals.insert(4, "Kitten")
print(myAnimals)
```

one useful thing is to create an empty list and add to it,

```python
myemptylist = []
print(myemptylist)
myemptylist.append(1)
print(myemptylist)
myemptylist.append(2)
print(myemptylist)
myemptylist.append(3)
print(myemptylist)
myemptylist.append(4)
print(myemptylist)
```

Once you have a list, remove every other item

```python
myList = []

for i in range(5):
  myList.append(input('Animal: '))

print("Original list:", myList)

# To remove every other item, we can slice the list to keep only even indices
myList = myList[::2]

print("After removing every other item:", myList)
```

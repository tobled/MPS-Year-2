### Loops
Loops are very important in computing, it allows code to perform the same process on a large amount of data or across a large number of variables.  We have already looked at loops, so I will keep this part brief.
There are two types of loops for and while loops

### For loops
For loops are great if you know how many times you need to loop. For example, if you need to process every item in a list or if you want count up to a specific number.  Here are some examples.

```python
import numpy as np

# Run through a list
myList = ['Cat', 'Dog', 'Hamster']
for currentAnimal in myList:
    print(currentAnimal)
```

```python
# Count to a number
for currentNumber in range(10):
    print(currentNumber)
```

```python
# The range function has some useful input conditions
for currentNumber in range(2,10,3): # Start at 2, finish at 10 and increment by 3
    print(currentNumber)
```

```python
# Yu can use any generator functions to provide you items to loop through
import numpy as np
for currentNumber in np.arange(2,10,3):
    print(currentNumber)
print(' ')

for currentNumber in np.linspace(2,10,3):
    print(currentNumber)
print(' ')

for currentNumber in np.logspace(2,10,3):
    print(currentNumber)
```

### While loops
While loops are great when you don't know in advance, how many loops you will have to perform. This is useful when performing iterative calculations.


```python
# here we can find the crossing point of a function
def myfunc(n):
    y = n + n**2 + n**3
    return y

diff = 100
xPos = -10
while diff > 1:
    y = myfunc(xPos)
    xPos = xPos + 0.1
    diff = np.abs(y)
print(xPos, y)
```

```python
# We do have to be careful with while loops as they can run for ever
# Best to have a safety net
def myfunc(n):
    y = n + n**2 + n**3
    return y

diff = 100
xPos = -10
safteyNet =0
while diff > 1:
    y = myfunc(xPos)
    xPos = xPos + 0.1
    diff = np.abs(y)
    safteyNet = safteyNet + 1
    if safteyNet > 1000:
        print('Loop broken')
        break
print(xPos, y)
```

```python
# You could also do this
def myfunc(n):
    y = n + n**2 + n**3
    return y

diff = 100
xPos = -10
safteyNet =0
while (diff > 1) and safteyNet < 1000:
    y = myfunc(xPos)
    xPos = xPos + 0.1
    diff = np.abs(y)
    safteyNet = safteyNet + 1
print(xPos, y)
```

## Enumerate

If we have a list that we would like to loop though, it could be a list of animals.


```python
myAnimals = ['Tiger', 'Snail', 'Iguana', 'Blowfish']
```

If I want to print the name and position of each animal I can do this

```python
for Position in range(len(myAnimals)):
    print('The ', myAnimals[Position], 'is at position ', Position + 1)
```

But I can use the enumerate command which makes better use of the loop

```python
for Position, Animal in enumerate(myAnimals):
    print('The ', Animal, 'is at position ', Position + 1)
```

```python
for Position, Animal in enumerate(myAnimals, 5): # I can define the start position for  enumerate
  print('The ', Animal, 'is at position ', Position)
```

Or less than, but how do we use this, well we can use the IF command

```python
if x < y:
    print('It is true')
```

```python
if x > y:
    print('It is true')
```

Here the IF command can be used to execute a command is a comparison is true, notice the syntax.  This is where python can get picky, after the IF statement there is a : this tells python it has a block of code to process. This block of code is defined by being indented, look at this

```python
x = 1
y = 10
if x == y:
    print("they are the same")
    print("what happens if you change one of the variables")
    print('Go on try it')
print("what is the worst that could happen")
```

We can make this more powerful in two different ways, the first is by joining conditions.

```python
x = 10
y = 11
z = 12
if x < y and z > y:
    print("all conditions met")
print("nothing to see here")
```

```python
x = 10
y = 11
z = 12
if x < y or x > z:
    print("all conditions met")
print("nothing to see here")
```

Here I have used the AND and OR operators to show that quite complex conditions can be explored.
We can also add complexity by having decision trees,

```python
x = 120
if x < 50:
    print('(first suite)')
    print('x is small')
else:
    print('(second suite)')
    print('x is large')
```

Here there are two blocks of code, one bellow the if statement which is executed if the condition is true and one bellow th else is the condition is not met. We can also add secondary if condition

```python
name = 'Joe'
if name == 'Fred':
    print('Hello Fred')
elif name == 'Xander':
    print('Hello Xander')
elif name == 'Joe':
    print('Hello Joe')
elif name == 'Arnold':
    print('Hello Arnold')
else:
    print("I don't know who you are!")
```

Here we have a series of if commands which are executed.

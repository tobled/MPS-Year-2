So far we have looked at using numbers both floats and integers.  Now we will look at strings.
As string is a collection of characters

```python
myString ='Hello'                                # Define a string

print(myString[0])                               # Print the first character
print(myString[1])                               # Print the second character
print(myString[2])                               # Print the third character
print()
new_str = myString[4] + myString[3] + myString[2] + myString[1] + myString[0]  # This is very different from numbers, here the + sign is adding a character to the end of the string
print(new_str)
#Notice that python starts counting at 0
```

A quick definition we will use the term character for a single letter and string for a word (a string of characters). Both can be stored in variables just as numbers.  Lets look at the above code in a little more detail.

```python
myNewString = "Bunny Kitten" # The computer has made a variable with 12 spaces and each space contains a character.
```

```python
print(len(myNewString)) # We can use the len() command to find out how much space the variable has
```

Slicing: Sounds very scary but it refers to process of selecting out parts of the string, i.e. accessing the individual pigeon holes containing the individual characters.

```python
print(myNewString[0]) # This prints the first character.  The [] brackets tell python we only want a small part of the object.
```

```python
print(myNewString[1]) # Prints the second number, remember that python starts counting at 0
```

```python
print(myNewString[0:5]) # we can select an range of characters
```

Note the second index specifies the last character that is not included in the result.

I can even skip a given number of characters

```python
print(myNewString[0:12:2]) # This prints every other character the third number is called the stride is skips one character each time
```

You can even count from the end of the string

```python
print(myNewString[-5:-1]) # This prints the last 5 to the last but one character
```

```python
print(myNewString[-5:]) # This prints the last 5 to the last character
```

Now your turn, write some code to print every other character in your full name.

```python
name = 'Tobias Sung'

for i in range(len(name)):
  print(name[i])
```

Can you work out how to print your name backwards using slicing

```python
temp = []
for i in range(len(name)):
  temp.append(name[-1*i-1])

print("".join(temp))
```
